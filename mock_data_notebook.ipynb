```python
type_mapping = {
    "int64": "BIGINT",
    "int32": "INT",
    "float64": "DOUBLE",
    "bool": "TINYINT(1)",
    "datetime64[ns]": "DATETIME",
    "object": "VARCHAR(255)",  # safe default varchar length
    "str": "VARCHAR(255)",
}
```

```python
column_defs = [f"`{col}` {type_mapping[str(dtype)]}" for col, dtype in df.dtypes.items()]
schema_sql = f"CREATE TABLE IF NOT EXISTS mock ({', '.join(column_defs)});"

schema_sql
```
